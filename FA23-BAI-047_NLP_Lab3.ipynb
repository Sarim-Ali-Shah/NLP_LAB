# Task 1: Regex Fundamentals (Character Classes & Quantifiers)

````python
import re

text = "Ali has 25 apples and 1500 mangoes. Karachi and Lahore are cities. Year 2023 was great. Cat dog sun moon programming beautiful"
````

````python
digits = re.findall(r'\d+', text)
print(digits)
````

````python
capital_words = re.findall(r'\b[A-Z][a-z]*\b', text)
print(capital_words)
````

````python
four_digits = re.findall(r'\b\d{4}\b', text)
print(four_digits)
````

````python
words_3_6 = re.findall(r'\b\w{3,6}\b', text)
print(words_3_6)
````

# Task 2: Pattern-Based Extraction (Real World Data)

````python
import re

text2 = "Contact ali@example.com or sara.khan@test.org. Call 0301-1234567 or 0333-9876543. Order ID: ORD12345 placed on 15/08/2023 and 01/01/2024. Discount is 45% and 12.5% off."
````

````python
emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text2)
print(emails)
````

````python
phones = re.findall(r'03\d{2}-\d{7}', text2)
print(phones)
````

````python
dates = re.findall(r'\d{2}/\d{2}/\d{4}', text2)
print(dates)
````

````python
percentages = re.findall(r'\d+\.?\d*%', text2)
print(percentages)
````

# Task 3: Word Tokenization (NLTK vs spaCy)


````python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
import spacy

paragraph = "I don't like waiting in long lines. It's really frustrating, isn't it? Dr. Smith said he'll arrive at 5 p.m. sharp. Let's hope the weather stays nice."
````

````python
nltk_tokens = word_tokenize(paragraph)
print(nltk_tokens)
````

````python
nlp_blank = spacy.blank("en")
doc = nlp_blank(paragraph)
spacy_tokens = [token.text for token in doc]
print(spacy_tokens)
````

Differences observed:
1. NLTK splits contractions like "don't" into "do" and "n't", spaCy blank pipeline keeps them split differently (e.g. "do" + "n't" vs different rule for "It's").
2. NLTK and spaCy handle punctuation attached to abbreviations like "p.m." differently, producing different token counts.

````python
print("NLTK token count:", len(nltk_tokens))
print("spaCy token count:", len(spacy_tokens))
````

# Task 4: Token Attributes (Categorizing Tokens)

````python
import spacy

nlp = spacy.load("en_core_web_sm")
sentence = "Ali paid $50 and Rs. 500 for the order, contact him at ali@example.com or call 03001234567."
doc = nlp(sentence)
````

````python
for token in doc:
    print(token.text, token.is_alpha)
````

````python
for token in doc:
    print(token.text, token.like_num, token.like_email)
````

````python
currency_tokens = [token.text for token in doc if token.is_currency]
print(currency_tokens)
````

# Task 5: Sentence Segmentation (Handling Tricky Cases)

````python
import spacy

nlp = spacy.load("en_core_web_sm")
paragraph2 = "Dr. Ahmed went to the market. He bought fruits, e.g. apples and bananas. Mr. Khan joined him later. They all had a great time."
doc2 = nlp(paragraph2)
````

````python
sentences = list(doc2.sents)
for sent in sentences:
    print(sent.text)
````

````python
print("Total sentences:", len(sentences))
````

Observation: "Dr." and "e.g." did not cause incorrect sentence breaks, spaCy correctly kept them within their sentences.
