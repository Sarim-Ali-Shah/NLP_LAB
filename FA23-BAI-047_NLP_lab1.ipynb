```python
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]
print("\n2. Token Validation:")
for t in tokens:
    if t.isalpha():
        status = "purely alphabetic"
    elif t.isdigit():
        status = "purely numeric"
    else:
        status = "mixed"
    print(f"   Token: '{t}' -> {status}")
```

```python
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]
pure_words = []
pure_numbers = []
mixed_tokens = []
for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)
print("\n3. Categorized Tokens:")
print("   Pure words:", pure_words)
print("   Pure numbers:", pure_numbers)
print("   Mixed tokens:", mixed_tokens)
```

```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

test_list = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]
print("\n4. Validation Function:")
for item in test_list:
    print(f"   {item} -> {validate_token(item)}")
```

### TASK 8

```python
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

test_str = "  HELLO Natural Language Processing  "
print("1. Clean Functions:")
print("   Regular function:", clean(test_str))
print("   Lambda function:", clean_lambda(test_str))
```

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence = "I love working with NLP tools in Python daily"
cleaned_sentence = clean(sentence)
tokens_list = tokenize(cleaned_sentence)
filtered_tokens = remove_short(tokens_list, 3)
print("\n2. Chained Functions:")
print("   Result:", filtered_tokens)
```

```python
def word_lengths(tokens):
    return {t: len(t) for t in tokens}

test_tokens = ["language", "model", "ai", "processing", "deep"]
lengths_dict = word_lengths(test_tokens)
print("\n3. Word Lengths:")
print("   Dictionary:", lengths_dict)
```

```python
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    long_tokens = remove_short(tokens, min_len)
    return word_lengths(long_tokens)

sent1 = "AI is transforming our everyday tasks dynamically"
sent2 = "Deep learning algorithms require substantial computation power"
print("\n4. NLP Pipeline:")
print("   Pipeline output 1:", nlp_pipeline(sent1, 3))
print("   Pipeline output 2:", nlp_pipeline(sent2, 4))
```
